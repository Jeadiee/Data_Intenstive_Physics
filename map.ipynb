# Map of uruguay 

```python
import ee
import geemap

ee.Initialize(project="datintensivephysics")

countries = ee.FeatureCollection(
    "FAO/GAUL_SIMPLIFIED_500m/2015/level0"
)

uruguay = countries.filter(
    ee.Filter.eq("ADM0_NAME", "Uruguay")
)

# geemap uses bokeh to display maps in Jupyter notebooks. Same as the talk we had 
Map = geemap.Map()

# Center the map on Uruguay with a zoom level of 6 (trial and error to find a good zoom level)
Map.centerObject(uruguay, 6)

# plots boundary of Uruguay in red and adds it to the map
Map.addLayer(
    uruguay,
    {"color": "red"},
    "Uruguay boundary"
)
# plots Sentinel-2 imagery for Uruguay in true color (RGB) and adds it to the map
Map.addLayer(
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(uruguay)
    .filterDate("2024-01-01", "2024-12-31")
    .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 20)) # Filter out cloudy pixels
    .median() 
    .clip(uruguay), #clip the image to the boundary of Uruguay
    {
        "bands": ["B4", "B3", "B2"],   # RGB bands for true color
        "min": 0,                      # Minimum reflectance value for visualization
        "max": 3000                     # Maximum reflectance value for visualization
    },
    "Sentinel-2 imagery"                # this gives it satellite style visuals. 
)

Map
```
